```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
print(sns.get_dataset_names())
```

```python
df = sns.load_dataset('penguins')
```

```python
sns.histplot(data=df, x='bill_length_mm', hue='species', multiple='stack')
plt.title('Distribution of Bill Length by Species')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Count')
```

```python
sns.boxplot(data=df, x='species', y='bill_length_mm')
plt.title('Boxplot of Bill Length by Species')
plt.xlabel('Species')
plt.ylabel('Bill Length (mm)')
```

```python
sns.countplot(data=df, x='species', hue='sex', palette='Set2')
```

```python
df.species.value_counts().plot(kind='bar')
```

```python
sns.pairplot(df, hue='species', height=2.5)
```

```python
sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species', style='sex', palette='Set1')
```

```python
df.info()
```

```python
sns.jointplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species')
```

```python
corr=df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Numerical Features')
```
